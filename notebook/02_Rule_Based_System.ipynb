```python
def rule_based_filter(data, aktivitas, undertone, hijab):
    """
    Filter outfit berdasarkan:
    - aktivitas
    - undertone
    - hijab
    """

    filtered = data.copy()

    filtered = filtered[
        filtered["aktivitas"] == aktivitas
    ]

    filtered = filtered[
        filtered["undertone"] == undertone
    ]

    filtered = filtered[
        filtered["hijab"] == hijab
    ]

    return filtered.reset_index(drop=True)
```

```python
candidate = rule_based_filter(
    metadata,
    aktivitas="dinner",
    undertone="warm",
    hijab="ya"
)

candidate.head()
```
